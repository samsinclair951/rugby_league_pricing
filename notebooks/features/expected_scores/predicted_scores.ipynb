```python
import pandas as pd

from rugby_league_pricing.database.connection import get_connection

query = """
SELECT
    f.match_date,
    home.canonical_name AS home_team,
    away.canonical_name AS away_team,
    ROUND(p.expected_home_score, 2) AS expected_home_score,
    ROUND(p.expected_away_score, 2) AS expected_away_score
FROM expected_score_predictions AS p
JOIN fixtures AS f
    ON p.fixture_id = f.fixture_id
JOIN teams AS home
    ON p.home_team_id = home.team_id
JOIN teams AS away
    ON p.away_team_id = away.team_id
ORDER BY
    f.match_date,
    home_team;
"""

with get_connection() as connection:
    predicted_scores = pd.read_sql_query(
        query,
        connection,
        parse_dates=["match_date"],
    )

predicted_scores
```
