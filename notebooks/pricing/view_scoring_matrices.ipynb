```python
import io

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from rugby_league_pricing.database.connection import get_connection

with get_connection() as connection:
    blob = connection.execute(
        """
        SELECT probability_matrix
        FROM historical_score_matrices
        ORDER BY as_of_date DESC
        LIMIT 1
        """
    ).fetchone()[0]
```

```python
matrix = np.load(io.BytesIO(blob), allow_pickle=False)

df = pd.DataFrame(matrix)

df.sample(5)
```

```python
def plot_score_matrix(
    matrix: np.ndarray,
    *,
    bucket: int = 2,
    max_score: int = 60,
) -> np.ndarray:
    """Aggregate and plot a score-probability matrix in score buckets."""
    if bucket <= 0:
        raise ValueError("bucket must be greater than zero.")

    if matrix.ndim != 2:
        raise ValueError("matrix must be two-dimensional.")

    if matrix.shape[0] != matrix.shape[1]:
        raise ValueError("matrix must be square.")

    score_limit = min(max_score + 1, matrix.shape[0])

    # Include enough cells to complete the final bucket.
    padded_size = int(np.ceil(score_limit / bucket) * bucket)

    trimmed = matrix[:score_limit, :score_limit]

    padded = np.zeros((padded_size, padded_size))
    padded[: trimmed.shape[0], : trimmed.shape[1]] = trimmed

    bucketed_matrix = padded.reshape(
        padded_size // bucket,
        bucket,
        padded_size // bucket,
        bucket,
    ).sum(axis=(1, 3))

    bucket_labels = np.arange(0, padded_size, bucket)

    plt.figure(figsize=(10, 8))

    plt.imshow(
        bucketed_matrix,
        origin="lower",
        aspect="equal",
    )

    plt.xlabel("Away score")
    plt.ylabel("Home score")
    plt.title(f"Historical score matrix ({bucket}-point buckets)")

    tick_positions = np.arange(len(bucket_labels))

    plt.xticks(tick_positions, bucket_labels)
    plt.yticks(tick_positions, bucket_labels)

    plt.colorbar(label="Probability")
    plt.tight_layout()
    plt.show()

    return None
```

```python
plot_score_matrix(matrix, bucket=2, max_score=60)
```

```python
plot_score_matrix(matrix, bucket=4, max_score=40)
```
